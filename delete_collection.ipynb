```python
# delete all collection in a chromaDB instance
PERSIST_DIR = "chroma_db"
import chromadb
from chromadb.config import Settings
chroma_client = chromadb.PersistentClient(path=PERSIST_DIR, settings=Settings(allow_reset=True))
collections = chroma_client.list_collections()
for collection in collections:
    chroma_client.delete_collection(name=collection.name)
chroma_client.reset()  # this will delete all collections


from chromadb.api.client import SharedSystemClient

chroma_client._system.stop()
SharedSystemClient._identifier_to_system.pop(chroma_client._identifier, None)
```
