```python
# Diagnostic script - run to inspect storage folders
import os
import json
import chromadb

PERSIST_DIR = "chroma_db"
STORAGE_DIR = "storage"
KNOW_PATH = "knowledge_md"

print("PWD:", os.getcwd())
print("PERSIST_DIR exists:", os.path.exists(PERSIST_DIR))
print("STORAGE_DIR exists:", os.path.exists(STORAGE_DIR))
print("KNOW_PATH exists:", os.path.exists(KNOW_PATH))
print("Files in PERSIST_DIR:", os.listdir(PERSIST_DIR) if os.path.exists(PERSIST_DIR) else "n/a")
print("Files in STORAGE_DIR:", os.listdir(STORAGE_DIR) if os.path.exists(STORAGE_DIR) else "n/a")

# check llama-index json files
for fname in ("docstore.json","index_store.json"):
    path = os.path.join(STORAGE_DIR, fname)
    print(path, "=>", os.path.exists(path))

# check chroma collection (if chromadb used)
try:
    client = chromadb.PersistentClient(path=PERSIST_DIR)
    try:
        cols = client.list_collections()
        print("chroma collections:", cols)
    except Exception as e:
        print("chroma client ok but list_collections failed:", e)
except Exception as e:
    print("chroma PersistentClient failed:", e)
```
