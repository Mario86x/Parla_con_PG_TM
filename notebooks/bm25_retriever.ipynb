```python
from rank_bm25 import BM25Okapi

# Corpus di documenti
corpus = [
    "il data analyst definisce i KPI aziendali",
    "il data engineer implementa pipeline ETL per raccogliere i dati",
    "il data scientist sviluppa modelli predittivi"
]

# Tokenizza i documenti
tokenized_corpus = [doc.split(" ") for doc in corpus]

# Inizializza BM25
bm25 = BM25Okapi(tokenized_corpus)

# Query
query = "modelli di machine learning per il forecast"
tokenized_query = query.split(" ")

# Calcola i punteggi BM25
scores = bm25.get_scores(tokenized_query)

# Ordina i documenti per rilevanza
results = sorted(zip(corpus, scores), key=lambda x: x[1], reverse=True)
for doc, score in results:
    print(f"{score:.2f} -> {doc}")
```
