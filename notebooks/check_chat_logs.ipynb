```python
# Cell 1 - Load and analyze chat logs
import json

# Open file with UTF-8 encoding
with open('chat_logs.json', 'r', encoding='utf-8') as f:
    chat_logs = json.load(f)

# Print user IDs (keys)
print("User IDs in logs:")
print(list(chat_logs.keys()))

# Print summary for each user
for user_id, interactions in chat_logs.items():
    print(f"\nUser {user_id}:")
    print(f"Number of interactions: {len(interactions)}")
    if interactions:
        first_time = interactions[0]['timestamp']
        last_time = interactions[-1]['timestamp']
        print(f"First interaction: {first_time}")
        print(f"Last interaction: {last_time}")

```
