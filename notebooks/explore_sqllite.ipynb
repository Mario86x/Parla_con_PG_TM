```python
# show all tables in database under ../chroma_db/chroma.sqlite3
import sqlite3
import os
import sys
import pandas as pd
import sqlalchemy as sa
```

```python
engine = sa.create_engine('sqlite:///../chroma_db/chroma.sqlite3')
with engine.connect() as conn:
    result = conn.execute(sa.text("SELECT name FROM sqlite_master WHERE type='table';"))
    tables = result.fetchall()
    print("Tables in database:")
    for table in tables:
        print(f"- {table[0]}")
```

```python
# list top 5 rows of each table
with engine.connect() as conn:
    for table in tables:
        table_name = table[0]
        result = conn.execute(sa.text(f"SELECT * FROM {table_name} LIMIT 3;"))
        rows = result.fetchall()
        print(f"\nTop 3 rows of table '{table_name}':")
        for row in rows:
            print(row)
```
